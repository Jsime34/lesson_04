```
import pandas as pd
campaign = pd.read_csv('https://raw.githubusercontent.com/byui-cse/cse450-course/master/data/bank.csv')

# import polars as pl
# campaign = pl.read_csv('https://raw.githubusercontent.com/byui-cse/cse450-course/master/data/bank.csv', schema_overrides={'nr.employed': pl.Float64})
```

```
marital_map = {"unknown": 0, "single": 1, "divorced": 2, "married": 3}
campaign["marital_encoded"] = campaign["marital"].map(marital_map)
```

```
import matplotlib.pyplot as plt

labels = {0: "unknown", 1: "single", 2: "divorced", 3: "married"}
counts = campaign["marital_encoded"].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(counts.index, counts.values)
ax.set_xticks(counts.index)
ax.set_xticklabels([f"{i}\n{labels[i]}" for i in counts.index])
ax.set_xlabel("Marital status (encoded)")
ax.set_ylabel("Number of clients")
ax.set_title("Clients by marital status")
plt.show()
```
